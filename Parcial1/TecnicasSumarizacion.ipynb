```python
import pandas as pd

url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/titanic.csv"
df = pd.read_csv(url)
```

# Asimetría (Skewness)
### Que es???????
Mide la inclinación de los datos. Indica si la mayoría de los valores se concentran a la izquierda (valores bajos con cola a la derecha) o a la derecha (valores altos con cola a la izquierda) respecto al promedio.

### ¿Qué significa físicamente sobre la edad de los pasajeros en el barco?

Si el resultado es positivo, significa que hay una mayor cantidad de personas jovenes y adultos jovenes  
Si el resultado es negativo, significa que hay una mayor cantidad de personas mayores o de la tercera edad   
En caso de que hubiera estado mas cerca a 0, eso nos indicaria que hay una distribucion mas equilibrada en los datos  
En este caso especifico, el skewness nos da 0.38910... esto significando que la mayoria de los pasajeros eran jovenes

```python
edad = df['Age'].skew()

print("Asimetria de edad:", edad)
```

# Curtosis (Kurtosis)

### Que es???????
Mide qué tan puntiaguda es la distribución y la presencia de valores atípicos (outliers) extremos en las colas. A mayor curtosis, más picos pronunciados y valores extremos existen.

### ¿Qué significa este número sobre el "grosor" de las colas de la distribución? ¿Confirma esto la existencia de los outliers que encontramos hoy en clase?
Un valor de curtosis de 33.40 indica que la distribución es leptocúrtica con colas sumamente pesadas/gruesas. Esto confirma la presencia de outliers extremos, ya que demuestra que aunque la mayoría pagó tarifas bajas y concentradas, hay boletos atípicos con precios demasiado altos que estiran la cola de la distribución

```python
fare = df['Fare'].kurt()

print(f"Curtosis de la columna Fare: {fare:.4f}")
```

# Tablas Cruzadas (Cross-Tabulation)

### Que es???????
Matriz que cruza dos variables categóricas para contar sus frecuencias conjuntas y descubrir si existe una relación o patrón entre ellas (ej. clase de boleto vs. supervivencia).

### ¿Qué patrón logran visualizar rápidamente al cruzar ambas variables?

Existe una relacion de privilegio: la probabilidad de sobrevivir aumentaba significativamente con una mejor clase de boleto  
En 1ra clase sobrevivieron mas personas de las que murieron (tasa de supervivencia superior al 60%)  
En 2da clase el resultado fue intermedio y relativamente dividido (cerca del 47% de supervivencia)  
En 3ra clase ocurrio la mayor mortandad en terminos absolutos y porcentuales (aproximadamente el 76% de los pasajeros fallecio)

```python
tablaCruzada = pd.crosstab(df['Pclass'], df['Survived'], margins=True)
print(tablaCruzada)
```
