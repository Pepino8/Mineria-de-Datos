```python
import pandas as pd
import numpy as np
 
 
# Reemplaza con la URL de tu repositorio
url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/googleplaystore.csv"
df = pd.read_csv(url)
```

```python
#Pregunta A
df.isnull().sum()

#Es correcto borrar las filas que no tienen Android version ya que como son muy pocos datos,
#no afecta mucho al dataset, pero borrar 1500 datos ya afecta a los resultados que se pueden obtener
```

```python
#Pregunta B

print(len(df)) #Empieza con 10841
df = df.dropna(subset=['Android Ver', 'Type'])
print(len(df)) #Termina con 10837
```

```python
# Pregunta C
df['Rating'] = df['Rating'].fillna(df.groupby('Category')['Rating'].transform('median'))

print("Nulos restantes en Rating:", df['Rating'].isnull().sum())
```

```python
#Pregunta D
df['Size'] = df['Size'].replace('Varies with device', np.nan)

print("Valores NaN en Size:", df['Size'].isnull().sum())

#Es mejor usar NaN porque poner '0MB' meteria un valor falso que podria afectar las estadisticas que se pueden obtener con el dataset
#Si lo convertimos a NaN, se conoce formalmente como un dato que falta y que puede ser utilizado despues para calcular las estadisticas
```
