```python
import pandas as pd
import numpy as np

df = pd.read_csv("playstore_limpio.csv")
```

# Reto 1
### ¿Cuál es el peso promedio en Megabytes de las apps en la Play Store?
El peso promedio de las apps es de 20.41 Megabytes


```python
def string_movedor(val):
    if not isinstance(val, str):
        return np.nan
    
    val = val.strip()
    if val.endswith('M'):
        return float(val[:-1])
    elif val.endswith('k'):
        return float(val[:-1]) / 1024
    elif val == "Varies with device":
        return np.nan
    else:
        return np.nan
```

```python
df['Size'] = df['Size'].apply(string_movedor)

df['Size'].mean()
```
