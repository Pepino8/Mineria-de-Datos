```python
import pandas as pd
url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/googleplaystore.csv"
df = pd.read_csv(url)
```

# Pregunta A
### ¿Cuántos registros duplicados purgaron?
1181 Registros fueron purgados


```python
df.duplicated(subset=['App']).sum() #1181

df.drop_duplicates(subset=['App'], keep='first', inplace=True)

df.duplicated(subset=['App']).sum() #0
```

# Pregunta B
### ¿Cuál es el promedio real de descargas?
El promedio de desargas es de 7.7 millones de descargas

```python

df['Installs'] = df['Installs'].astype(str).str.replace('+', '', regex=False).str.replace(',', '', regex=False)

df['Installs'] = pd.to_numeric(df['Installs'], errors='coerce')

df['Installs'].mean()
```

# Pregunta C

```python
df['Price'] = df['Price'].astype(str).str.replace('$', '', regex=False)

df['Price'] = pd.to_numeric(df['Price'], errors='coerce')
```

# Pregunta D

```python
df['Price'].max()

df[df['Price'] > 200]

df = df[df['Price'] < 50]

df.to_csv('playstore_limpio.csv', index=False)
```
