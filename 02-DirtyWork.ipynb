```python
import random

random_list = [random.random() for _ in range(5_000_000)]

print("Random list length:", len(random_list))
```

```python
def getSD():
    sd = 0
    numerador = 0
    denominador = len(random_list)

    promedio = 0

    #sacar promedio
    for num in random_list:
        promedio += num
    
    promedio = promedio / denominador

    #sacar numerador
    for num in random_list:
        numerador += (num - promedio) ** 2

    sd = numerador / denominador

    sd = sd ** 0.5

    print(sd)

getSD()
```

```python
import numpy as np

arr = np.array(random_list)

print(f'Std: {arr.std()}')
```
