```python
import pandas as pd
 
 
url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/titanic.csv"
df = pd.read_csv(url)

df.head()

#df.describe()

promedio_M = 0
promedio_F = 0

m = 0
f = 0
for i, row in df.iterrows():
    sex = row['Sex']
    age = row['Age']

    if not pd.isna(age):
        if sex == "male":
            m += 1
            promedio_M += age
        else:
            f += 1
            promedio_F += age
    
promedio_M = promedio_M / m 
promedio_F = promedio_F / f 

print("Promedio Femenino:", promedio_F)
print("Promedio Masculino:", promedio_M)

df.groupby('Pclass')['Fare'].mean()

# Cambia 'data' por 'df':
print(df['Fare'].quantile(0.25))
print(df['Fare'].quantile(0.5))
print(df['Fare'].quantile(0.75))
print(df['Fare'].quantile(1))

```
