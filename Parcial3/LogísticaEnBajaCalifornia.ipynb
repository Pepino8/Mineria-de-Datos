# Práctica: Logística en Baja California

## Fase 1: Exploración Inicial

En esta primera fase se realiza:
1. **Carga de datos:** Importación del archivo `logistica_baja_california.csv` usando `pandas`.
2. **Escalamiento con `StandardScaler`:** Estandarización de las columnas `Latitud` y `Longitud` para generar la matriz `X_scaled`.  
   *Nota:* Un grado de longitud no mide exactamente los mismos kilómetros que un grado de latitud debido a la curvatura de la Tierra. Estandarizar nivela el terreno matemático.
3. **Visualización de datos crudos:** Gráfica de dispersión (*Scatter Plot*) con `Longitud` en el eje X y `Latitud` en el eje Y, revelando la silueta de la costa del Pacífico de Baja California.

### 1. Ingesta e Inspección de Datos
Importamos las librerías necesarias y cargamos el archivo CSV.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler

# 1. Importar el CSV usando pandas
df = pd.read_csv('logistica_baja_california.csv')

# Dimensiones y primeras filas
print(f"Dimensiones: {df.shape[0]} filas y {df.shape[1]} columnas\n")
df.head()
```

```python
# Información de columnas y tipos de datos
df.info()

# Estadísticas descriptivas de las coordenadas
df.describe()
```

### 2. Escalamiento con StandardScaler (`X_scaled`)
Utilizamos `StandardScaler` sobre las columnas `Latitud` y `Longitud` y guardamos la matriz escalada en `X_scaled`.

> **Recuerden:** Un grado de longitud no mide exactamente los mismos kilómetros que un grado de latitud. Escalar nivela el terreno matemático centrando las variables en media 0 y varianza unitaria (1).

```python
# Instanciar StandardScaler
scaler = StandardScaler()

# Ajustar y transformar sobre las columnas Latitud y Longitud
X_scaled = scaler.fit_transform(df[['Latitud', 'Longitud']])

# Verificación de la matriz escalada
print(f"Forma de X_scaled: {X_scaled.shape}")
print(f"Media por columna: {X_scaled.mean(axis=0).round(4)}")
print(f"Desviación estándar por columna: {X_scaled.std(axis=0).round(4)}\n")
print("Primeras 5 filas de X_scaled:")
print(X_scaled[:5])
```

### 3. Gráfica de Dispersión (Scatter Plot) de Datos Crudos
Graficamos los datos crudos con `Longitud` en el eje X y `Latitud` en el eje Y. La forma de los puntos delinea con claridad la silueta de la costa del Pacífico de Baja California (desde la zona metropolitana de Tijuana/Rosarito hasta Ensenada y hacia el sur).

```python
# Scatter plot con datos crudos (Longitud en X, Latitud en Y)
plt.figure(figsize=(9, 10))
plt.scatter(
    df['Longitud'], 
    df['Latitud'], 
    color='#1f77b4', 
    alpha=0.6, 
    edgecolors='none', 
    s=25
)

plt.title('Distribución Geográfica de Solicitudes\nSilueta de la Costa del Pacífico (Baja California)', fontsize=14, pad=12)
plt.xlabel('Longitud', fontsize=12)
plt.ylabel('Latitud', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

### Gráfica complementaria: Distribución por Tipo de Solicitud
Podemos visualizar la misma silueta costera distinguiendo el tipo de servicio logístico (`Tipo_Solicitud`).

```python
plt.figure(figsize=(10, 11))
sns.scatterplot(
    data=df, 
    x='Longitud', 
    y='Latitud', 
    hue='Tipo_Solicitud', 
    palette='tab10', 
    alpha=0.7, 
    s=30
)

plt.title('Solicitudes por Tipo en la Costa de Baja California', fontsize=14, pad=12)
plt.xlabel('Longitud', fontsize=12)
plt.ylabel('Latitud', fontsize=12)
plt.legend(title='Tipo de Solicitud', loc='upper right')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
