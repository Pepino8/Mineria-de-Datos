# Reconstruir la matriz

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. Definimos las dimensiones del modelo (Notación ISLP)
n = 5 # Número de observaciones (Usuarios)
p = 3 # Dimensionalidad (Features: Edad, Salario_USD, Engagement_Score)

# Construimos la Matriz X (n x p) con datos de juguete
X_matriz = np.array([
    [25, 3000, 8.5],  # Usuario 1: Vector en 3D
    [45, 8000, 4.2],  # Usuario 2
    [22, 1500, 9.8],  # Usuario 3
    [35, 5000, 6.1],  # Usuario 4
    [28, 4200, 7.5]   # Usuario 5
])

# Para comodidad humana, la envolvemos en un DataFrame
df_X = pd.DataFrame(X_matriz, columns=['Edad', 'Salario', 'Engagement'], 
                    index=['U1', 'U2', 'U3', 'U4', 'U5'])

print("Nuestra Matriz Multidimensional X (n=5, p=3):")
display(df_X)
```

```python
# Extrayendo los vectores de los usuarios
vector_u2 = X_matriz[1]
vector_u4 = X_matriz[3]

print(f"Coordenadas Usuario 1: {vector_u2}")
print(f"Coordenadas Usuario 3: {vector_u4}")

# Calculando la distancia euclideana multidimensional pura (Norma L2)
distancia = np.linalg.norm(vector_u2 - vector_u4)

print(f"\nLa distancia geométrica exacta entre estos dos usuarios es: {distancia:.2f} unidades.")
```

```python
# Fórmula: Z = (X - mu) / sigma
media = np.mean(X_matriz, axis=0)
desviacion = np.std(X_matriz, axis=0)  # ddof=0 (convención matemática y de Machine Learning)

# Matriz estandarizada
X_z = (X_matriz - media) / desviacion

# Envolvemos en DataFrame para visualizarla claramente
df_X_z = pd.DataFrame(X_z, columns=['Edad', 'Salario', 'Engagement'],
                      index=['U1', 'U2', 'U3', 'U4', 'U5'])

print("Matriz Estandarizada Z-Score:")
display(df_X_z)
```

```python
# 2. Extracción de vectores para el Usuario 2 (índice 1) y Usuario 4 (índice 3)
vector_u2 = X_z[1]
vector_u4 = X_z[3]

print(f"Coordenadas U2 (estandarizado): {vector_u2}")
print(f"Coordenadas U4 (estandarizado): {vector_u4}")

# 3. Distancia euclidiana sobre la matriz estandarizada (Norma L2)
distancia_z = np.linalg.norm(vector_u2 - vector_u4)

print(f"\nDistancia euclidiana entre U2 y U4 (estandarizada): {distancia_z:.4f} unidades")
```

Ahora esta mas estandarizada la grafica y se notan mas las diferencias entre ambas variables
